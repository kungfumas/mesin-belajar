```r
# Construct the raw GitHub URL
github_url <- "https://raw.githubusercontent.com/kungfumas/mesin-belajar/master/ipm2024.csv"

# Read the CSV file from the URL into a dataframe
dataipm <- read.csv(github_url)

# Display the first few rows of the dataframe
head(dataipm)
```

```r
mean_ipm <- mean(dataipm$IPM2024)
print(mean_ipm)
```

```r
# Function to calculate geometric mean
geometric_mean <- function(x) {
  prod(x)^(1/length(x))
}

# Calculate the geometric mean of IPM2024
geometric_mean_ipm <- geometric_mean(dataipm$IPM2024)
print(geometric_mean_ipm)
```

```r
# Function to calculate harmonic mean
harmonic_mean <- function(x) {
  1 / mean(1 / x)
}

# Calculate the harmonic mean of IPM2024
harmonic_mean_ipm <- harmonic_mean(dataipm$IPM2024)
print(harmonic_mean_ipm)
```

```r
median_ipm <- median(dataipm$IPM2024)
print(median_ipm)
```

```r
# Function to calculate the mode
calculate_mode <- function(x) {
  ux <- unique(x)
  ux[which.max(tabulate(match(x, ux)))]
}

# Calculate the mode of IPM2024
mode_ipm <- calculate_mode(dataipm$IPM2024)
print(mode_ipm)
```

```r
# Calculate the deciles of IPM2024
deciles_ipm <- quantile(dataipm$IPM2024, probs = seq(0.1, 0.9, by = 0.1))
print(deciles_ipm)
```

```r
# Install and load ggplot2 if you haven't already
# install.packages("ggplot2")
library(ggplot2)

# Create a bar plot of IPM2024 by Provinsi
ggplot(dataipm, aes(x = Provinsi, y = IPM2024)) +
  geom_bar(stat = "identity") +
  theme(axis.text.x = element_text(angle = 90, hjust = 1)) + # Rotate x-axis labels for readability
  labs(title = "IPM2024 by Provinsi",
       x = "Provinsi",
       y = "IPM2024")
```

```r
# Calculate the range of IPM2024
range_ipm <- max(dataipm$IPM2024) - min(dataipm$IPM2024)
print(range_ipm)
```

```r
# Calculate the Mean Absolute Deviation (Simpangan Rata-Rata) of IPM2024
mean_absolute_deviation_ipm <- mean(abs(dataipm$IPM2024 - mean(dataipm$IPM2024)))
print(mean_absolute_deviation_ipm)
```

```r
# Calculate the variance of IPM2024
variance_ipm <- var(dataipm$IPM2024)
print(variance_ipm)
```

```r
# Calculate the standard deviation of IPM2024
standard_deviation_ipm <- sd(dataipm$IPM2024)
print(standard_deviation_ipm)
```
